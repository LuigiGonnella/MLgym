# 🟢 Easy: Implement ReLU

Implement the **ReLU** (Rectified Linear Unit) activation function from scratch.

$$\text{ReLU}(x) = \max(0, x)$$

### Signature
```python
def relu(x: torch.Tensor) -> torch.Tensor:
    ...
```

### Rules
- Do **NOT** use `torch.relu`, `F.relu`, `torch.clamp`, or any built-in activation
- Must support autograd (gradients should flow back)

### Example
```
Input:  tensor([-2., -1., 0., 1., 2.])
Output: tensor([ 0.,  0., 0., 1., 2.])
```

````python
# ✏️ YOUR IMPLEMENTATION HERE

def relu(x: torch.Tensor) -> torch.Tensor:
     #if we return torch.max(torch.zeros_like(x), x) --> if x > 0 then derivative is 1, if x < 0 then derivative is 0, but if x == 0 torch.max would split the derivatives between the branch `0` and the branch `x`
     #meaning that in y = max(a, b) if a == b --> dy/da = dy/db = 0.5
     #but we want the derivative of x being 0 when x == 0 (relu is not derivable in x == 0 but convention want drelu/dx = 0)
     #so we explicitly state that with the where function, which takes the `x` value if x > 0 and takes 0 otherwise
     return torch.where(x > 0, x, 0)
````

````python
# 🧪 Test your implementation (feel free to add more debug prints)
x = torch.tensor([-2., -1., 0., 1., 2.])
print("Input: ", x)
print("Output:", relu(x))
print("Shape: ", relu(x).shape)
````
