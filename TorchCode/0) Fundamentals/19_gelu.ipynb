# 🟢 Easy: GELU Activation

Implement the **GELU** (Gaussian Error Linear Unit) activation.

$$\text{GELU}(x) = x \cdot \Phi(x) = x \cdot 0.5 \cdot (1 + \text{erf}(x / \sqrt{2}))$$

### Signature
```python
def my_gelu(x: Tensor) -> Tensor: ...
```

### Rules
- Do NOT use `F.gelu`, `nn.GELU`, or `torch.nn.functional.gelu`
- Use `torch.erf` for the exact version

````python
import torch
import math
````

````python
# ✏️ YOUR IMPLEMENTATION HERE

def my_gelu(x):
  #gelu multiplies X for the Cumulative Density Function of a Normal Z ~ N(0, 1) w.r.t. --> P(Z <= x)
  #the result is integral between -inf and x of the probability density function (gaussian with mean = 0 and var = 1) but it has no solution in closed form
  #so this integral is approximated to 0.5 + (1 + erf(x/sqrt(2))
  #erf(x) = 2/pi * integral from 0 to x of exp(-t(2))dt called `error function`

  #--> anyway, if x very negative --> gate (CDF) ~ 0, if x almost zero --> gate with a very low value, if x very positive --> x practically unchanged
  #very similar to what ReLU does, but negative activations are not flattened to zero.
    return x * 0.5 * (1 + torch.erf(x/(2 ** 0.5)))
````

````python
# 🧪 Debug
x = torch.tensor([-2., -1., 0., 1., 2.])
print('Output:', my_gelu(x))
print('Ref:   ', torch.nn.functional.gelu(x))
````
